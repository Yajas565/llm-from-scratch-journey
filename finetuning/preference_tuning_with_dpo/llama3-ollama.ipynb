```python
import json
import requests

def query_model(prompt, model="llama3", url="http://localhost:11434/api/chat", role="user"):
    data = {
        "model": model,
        "seed": 123,
        "temperature": 1.,
        "top_p": 1,
        "messages": [
            {"role":role, "content": prompt}
        ]
    }

    with requests.post(url, json=data, stream=True, timeout=40) as r:
        r.raise_for_status()
        response_data = ""
        for line in r.iter_lines(decode_unicode=True):
            if not line:
                continue
            data = json.loads(line)
            if "message" in data:
                response_data += data['message']['content']
    return response_data
```

```python
def extract_instruction(text):
    for content in text.split('\n'):
        if content:
            return content.strip()
```

```python
query = "<|begin_of_text|><|start_header_id|>user<|end_header_id|>"
result = query_model(query, role="assistant")
result
```

```python
instruction = extract_instruction(result)
instruction
```

```python
response = query_model(instruction)
response
```
